# ⚙️ Manipulation de fichiers avec Python
---

Pour manipuler des fichiers et des répertoires, on utilise le module **os** (_operating system_). Il fournit une interface portable entre Python et le système d'exploitation (Windows, macOS, Linux).

Il permet :

📌 d'interagir avec le système de fichiers (fichiers, dossiers, chemins),\
📌 d'interagir avec les variables d'environnement,\
📌 d'exécuter des commandes système,\
📌 d'obtenir des informations sur le système (nom, process, etc.).

**🔑 Point clé :** Grâce à **os**, le même code Python fonctionne sur des OS différents — le module s'occupe des différences internes.


```python
# importation du module os
import os  #os : operating system
```

## 2️⃣ Lecture et écriture dans un fichier

À l'aide d'un éditeur de texte sans mise en forme, créer dans le répertoire de travail courant un fichier nommé fichier.txt contenant une simple phrase.

#### 2.1. Ouverture du fichier  
On utilise la fonction **open** (disponible sans import d'un quelconque module). La fonction prend en paramètres :  
* le chemin d'accès au fichier (absolu ou relatif)
* le mode d'ouverture (donné sous la forme d'une chaîne de caractères)

##### Modes d'ouverture les plus courants
* "r" : en lecture (_read_)
* "w" : en écriture (_write_) avec écrasement du contenu déjà existant ; création du fichier s'il n'existe pas
* "a" : en écriture en mode ajout (_append_), le nouveau contenu est ajouté à la fin du fichier sans écraser le contenu existant ; création du fichier s'il n'existe pas.

```python
mon_essai=open("mon_fichier.txt","r")
mon_essai
```

```python
print(type(mon_essai))
```


##### 💡 Quelques explications

"_io.TextIOWrapper" est la **classe** que Python utilise en coulisses pour représenter **un fichier texte ouvert**. La variable mon_essai n'est pas un simple fichier, c'est un **objet de la classe _io.TextIOWrapper**.

Décomposition du nom :

* _io : module interne (input/output) qui gère les entrées/sorties.
* TextIO : entrées/sorties de texte (par opposition aux octets).
* Wrapper : "enveloppe". L'objet enveloppe un flux d'octets bruts (un FileIO) pour permettre de lire et écrire du texte.

**Analogie**

Un fichier sur le disque ne contient que des octets (des nombres). TextIOWrapper est comme un traducteur attaché au fichier : il transforme ces octets en chaînes de caractères (à la lecture) et inversement (à l'écriture), grâce à un encodage (UTF-8, Latin-1...).

#### 2.2. Fermeture du fichier
❗️❗️ Ne pas oublier

On utilise la méthode **close()**.

```python
mon_essai.close()
```

#### 2.3. Lecture du fichier  
On utilise la méthode **read()**.

```python
mon_essai=open("mon_fichier.txt","r") #on ouvre le fichier en lecture
contenu=mon_essai.read()
print(contenu)
```

```python
mon_essai.close() #on ferme le fichier en lecture pour l'ouvrir ensuite en écriture
```

```python
mon_essai=open("mon_fichier.txt","a")
mon_essai.write("\nUne ligne de plus dans mon fichier")
mon_essai.close() # on ferme le fichier en écriture avant de l'ouvrir en lecture
mon_essai=open("mon_fichier.txt","r")
contenu=mon_essai.read()
print(contenu)
```

```python
mon_essai.close()
```

## 3️⃣ Enregistrer un objet dans un fichier

On peut enregistrer n'importe quel objet dans un fichier grâce au module **pickle**. On peut ensuite le récupérer, au lancement du programme par exemple. Ce module permet l'utilisation des deux classes **Pickler** et **Unpickler**.

#### 3.1. Enregistrement de l'objet

```python
# import du module pickle
import pickle
```

```python
os.listdir (".")
```

```python
# enregistrement d'un objet dans le fichier : appel de la méthode dump
coefficients_UE1={
    "R101": 10,
    "R102": 10,
    "R103": 7,
}
with open('donnees','wb') as fichier: #le fichier sera fermé automatiquement
    mon_objet_pickler=pickle.Pickler(fichier)
    mon_objet_pickler.dump(coefficients_UE1)
```

```python
os.listdir (".")
```

#### 3.2. Récupération de l'objet

```python
# lecture d'un objet dans un fichier : appel de la méthode load
with open('donnees','rb') as fichier:
    mon_objet_unpickler=pickle.Unpickler(fichier)
    coefficients_recuperes=mon_objet_unpickler.load()
```

```python
print(coefficients_recuperes)
```
