```python
import matplotlib.pyplot as plt
import numpy as np

def calcul_moyenne():
   c = 0
   note1 = 0
   note2 = 0
   note3 = 0
   ue1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
   ue2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
   ue3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]
   note = {"r101":0,"r102":0,"r103":0,"r104":0,"r105":0,"r106":0,"r107":0,"r108":0,"r109":0,"r110":0,"r111":0,"r112":0,"r113":0,"r114":0,"r115":0,"sae11":0,"sae12":0,"sae13":0,"sae14":0,"sae15":0,"sae16":0}
   for cle in note:
       print("Pour la resource",cle)
       note[cle] = int(input("note :"))
       while note[cle] < 0 or note[cle] > 20 :
           note[cle] = int(input("Note non comprise entre 0 et 20, merci de mettre une note valable pour la resource :"))
   for i in note :
       note1 += int(note[i])*ue1[c]
       note2 += int(note[i])*ue2[c]
       note3 += int(note[i])*ue3[c]
       c+=1
   note1 = note1/113
   note2 = note2/89
   note3 = note3/98
   return note1,note2,note3

def graph():
   n1, n2, n3 = calcul_moyenne()
   cate = [f"UE1 {n1:05.3f}", f"UE2 {n2:05.3f}", f"UE3 {n3:05.3f}"]
   val = [n1, n2, n3]
   colors = []
   for note in val:
       if note < 8:
           colors.append('red')
           
       elif 8 <= note < 10:
           colors.append('orange')
       else:
           colors.append('green')
   plt.bar(cate, val, color=colors)
   plt.title("Notes par UE")
   plt.ylim(0, 20)
   plt.show()


graph()

```
