```python
from tkinter import *
from tkinter import ttk

fenetre=Tk()
#taille de fenetre
fenetre.geometry("800x600")

frm=ttk.Frame(fenetre)
frm.grid()

ress=[f"R{i}" for i in range(101,116)]
sae=[f"SAE{i}" for i in range(11,17)]

notes = []

def notes_(c,liste):

   for i, matiere in enumerate(liste):
   
       etiq=ttk.Label(frm, text=matiere)
       etiq.grid(column=c, row=i, padx=10, pady=10)
       saisie=ttk.Entry(frm)
       saisie.grid(column=c+1, row=i, padx=10, pady=10)
       notes.append(saisie)

       
notes_(0,ress)
notes_(2,sae)

def recup():
   saisies= [i.get() for i in notes]
   
btn1 = ttk.Button(frm, text="Valider", command=notes_(0,ress))
btn1.grid(column = 3, row=8)


#savoir faire un ascenseur pour derouler la fen
fenetre.mainloop()
```
