```python
def calcul_moyenne():
   c = 0
   note1 = 0
   note2 = 0
   note3 = 0
   ue1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
   ue2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
   ue3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]
   note = []
   for i in range(len(ue1)):
       note.append(int(input("note :")))
   for i in note :
       note1 += i*ue1[c]
       note2 += i*ue2[c]
       note3 += i*ue3[c]
       c+=1
   note1 = note1/115
   note2 = note2/89
   note3 = note3/98
   return note1,note2,note3

calcul_moyenne()

```
