```python
def calcul_moyenne():
   c = int(input("Combien il y a de resource"))
   note1 = 0
   note2 = 0
   note3 = 0
   ue1 = 0
   ue2 = 0
   ue3 = 0
   for i in range(c) :
       x = int(input("note"))
       coef1 = int(input("coef pour l'UE1"))
       coef2 = int(input("coef pour l'UE2"))
       coef3 = int(input("coef pour l'UE3"))
       note1 += x*coef1
       note2 += x*coef2
       note3 += x*coef3
       ue1 += coef1
       ue2 += coef2
       ue3 += coef3
   note1 = note1/ue1
   note2 = note2/ue2
   note3 = note3/ue3
   return note1,note2,note3

calcul_moyenne()

```
